```python
import pandas as pd
import glob

files = sorted(glob.glob('../data/part*.csv'))
for f in files:
    print(f)
```

```python
df = pd.concat([pd.read_csv(f) for f in files], ignore_index=True)
print(df.shape)
df.head()
```

```python
for col in ['category', 'question', 'answer']:
    df[col] = df[col].str.strip()

print("Missing values:")
print(df.isnull().sum())

df = df.dropna(subset=['question', 'answer'])

df['q_key'] = df['question'].str.lower()
dups = df[df.duplicated('q_key', keep=False)].sort_values('q_key')
print("\nDuplicate questions:", len(dups))
dups[['category', 'question']]
```

```python
df = df.drop_duplicates(subset='q_key').drop(columns='q_key')
df = df.reset_index(drop=True)
df.insert(0, 'id', df.index + 1)

df.to_csv('../data/python_qa_clean.csv', index=False)
print(df.shape)
print(df['category'].value_counts())
```
